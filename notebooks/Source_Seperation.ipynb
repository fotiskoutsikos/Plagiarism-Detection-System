```python
import os
import glob
import subprocess

# Folder paths in Drive
audio_folder = "/content/drive/MyDrive/Plagiarism-Detection-System/data/segment_smp/audio/"
stems_output_dir = "/content/drive/MyDrive/Plagiarism-Detection-System/data/separated_segment_smp/"
os.makedirs(stems_output_dir, exist_ok=True)

# Load all files
test_files = glob.glob(os.path.join(audio_folder, "*.wav"))
print(f"Found {len(test_files)} files for seperation.")

def separate_all_stems(audio_input_path, out_dir):
    model_name = "mdx_extra_q"
    base_name = os.path.splitext(os.path.basename(audio_input_path))[0]
    final_stem_dir = os.path.join(out_dir, model_name, base_name)

    # Skip if exists
    expected_stems = ['vocals.wav', 'bass.wav', 'drums.wav', 'other.wav']
    if os.path.exists(final_stem_dir):
        existing_files = os.listdir(final_stem_dir)
        if all(stem in existing_files for stem in expected_stems):
            print(f"File {base_name} is already seperated. Ignoring...")
            return final_stem_dir

    print(f"Seperating: {base_name} ...")
    cmd = f'demucs -n {model_name} "{audio_input_path}" -o "{out_dir}"'

    try:
        subprocess.run(cmd, shell=True, check=True, capture_output=True, text=True)
    except subprocess.CalledProcessError as e:
        print(f"Error seperating {base_name}:\n{e.stderr}")
        return None

    return final_stem_dir

# Main Loop
print("\nStarting source seperation...")
for audio_file in test_files:
    separate_all_stems(audio_file, stems_output_dir)

print(f"\nSeperation Completed Successfully! All files are located at {stems_output_dir}")
```
