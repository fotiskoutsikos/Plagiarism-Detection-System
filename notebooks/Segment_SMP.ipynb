```python
import os
import pandas as pd
import ast
import librosa
import soundfile as sf

def create_segments_dataset():
    base_dir = '/content/drive/MyDrive/Plagiarism-Detection-System'
    old_csv_path = os.path.join(base_dir, 'Final_dataset_pairs.csv')
    old_dataset_dir = os.path.join(base_dir, 'data', 'final_dataset')

    new_dataset_dir = os.path.join(base_dir, 'data', 'segment_smp')
    audio_out_dir = os.path.join(new_dataset_dir, 'audio')
    new_csv_path = os.path.join(new_dataset_dir, 'segments_metadata.csv')

    os.makedirs(audio_out_dir, exist_ok=True)

    df = pd.read_csv(old_csv_path)

    metadata_records = []

    print(f"Starting SMP segmentation at: {new_dataset_dir}")

    for index, row in df.iterrows():
        pair_num = row['pair_number']

        pair_dir = os.path.join(old_dataset_dir, str(pair_num))
        ori_dir = os.path.join(pair_dir, 'original')
        comp_dir = os.path.join(pair_dir, 'comp')

        try:
            ori_file = os.path.join(ori_dir, os.listdir(ori_dir)[0])
            comp_file = os.path.join(comp_dir, os.listdir(comp_dir)[0])
        except (FileNotFoundError, IndexError):
            print(f"Ignored pair {pair_num} - No audio files found.")
            continue

        try:
            ori_times = ast.literal_eval(row['ori_times'])
            comp_times = ast.literal_eval(row['comp_times'])
        except (ValueError, SyntaxError):
            print(f"Error at timestamps of pair {pair_num}. Ignored.")
            continue

        for start_time in ori_times:
            end_time = start_time + 20
            out_filename = f"pair_{pair_num}_ori_{start_time}s.wav"
            out_filepath = os.path.join(audio_out_dir, out_filename)

            if not os.path.exists(out_filepath):
                try:
                    y, sr = librosa.load(ori_file, offset=start_time, duration=20)
                    sf.write(out_filepath, y, sr)
                except Exception as e:
                    print(f"Error at segmentation {out_filename}: {e}")
                    continue

            metadata_records.append({
                'segment_filename': out_filename,
                'pair_number': pair_num,
                'type': 'original',
                'start_time': start_time,
                'end_time': end_time,
                'title': row['ori_title'],
                'relation': row['relation']
            })

        for start_time in comp_times:
            end_time = start_time + 20
            out_filename = f"pair_{pair_num}_comp_{start_time}s.wav"
            out_filepath = os.path.join(audio_out_dir, out_filename)

            if not os.path.exists(out_filepath):
                try:
                    y, sr = librosa.load(comp_file, offset=start_time, duration=20)
                    sf.write(out_filepath, y, sr)
                except Exception as e:
                    print(f"Error at segmentation {out_filename}: {e}")
                    continue

            metadata_records.append({
                'segment_filename': out_filename,
                'pair_number': pair_num,
                'type': 'comp',
                'start_time': start_time,
                'end_time': end_time,
                'title': row['comp_title'],
                'relation': row['relation']
            })

    new_df = pd.DataFrame(metadata_records)

    new_df = new_df.drop_duplicates(subset=['segment_filename'])

    new_df.to_csv(new_csv_path, index=False)
    print(f"\nCompleted! Created {len(new_df)} segments.")
    print(f"Saved metadata CSV at: {new_csv_path}")

if __name__ == "__main__":
    create_segments_dataset()
```
